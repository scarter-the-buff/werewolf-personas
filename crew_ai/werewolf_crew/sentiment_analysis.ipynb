# Sentiment Analysis

We run sentiment analysis on agent utterances to determine the effect of personality on what the agents say.

```python
import nltk

nltk.download('all')
```

```python
# Preprocessing

import re

path = "./logs/2025-04-25_07-54-38_round_1_output .txt"

with open(path, "r") as f:
    lines = f.readlines()
    transcript = str(lines[1])


trans = re.sub(r'\\n', '\n', transcript)
trans = re.sub(r'\\', '', trans)

print(trans)
# Process out 
```

```python
from collections import defaultdict

lines_by_character = defaultdict(list)

pattern = re.compile(r'^([A-Za-z]+):\s*(.*)$')

for line in trans.split('\n'):
    match = pattern.match(line)
    if match:
        speaker = match.group(1)
        speech = match.group(2)
        lines_by_character[speaker].append(speech)

lines_by_character = dict(lines_by_character)

for character, lines in lines_by_character.items():
    print(f"{character}:")
    for l in lines:
        print(f"  - {l}")
    print()
```

```python
import nltk
from nltk.sentiment import SentimentIntensityAnalyzer


utterance = trans

sia = SentimentIntensityAnalyzer()

scores = sia.polarity_scores(utterance)


for character, lines in lines_by_character.items():
    print(f"{character}:", end=" ")
    print(scores['pos'] - scores['neg'])

```
